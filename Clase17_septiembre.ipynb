```python
#RC4
## cifrador de flujo XOR weP
from Crypto.Cipher import ARC4
clave = b"CLave1234"
mensaje = "Hoy es jueves"
cifrador = ARC4.new(clave)
mensaje_cifrado = cifrador.encrypt(mensaje.encode("utf-8"))
print("mensaje Original", mensaje)
print("mensaje Cifrado", mensaje_cifrado)

```

```python
descifrador = ARC4.new(clave)
mensaje_descifrado= descifrador.decrypt(mensaje_cifrado)

print("Mensaje descrifrado" , mensaje_descifrado.decode("utf-8"))
```

```python
## RC5 y RC6
## suma modular , XOR, rotacion
def rotacionIz(valor, cantidad):
  return ((valor << cantidad) |
          (valor >> (8- cantidad))) & 0xFF

A = 100
B = 200
clave = 50

print("A =" , A)
print("B =" , B)
print("Clave=" , clave)
```

```python
# suma
A = (A + clave) % 256
print ("A = " , A)

#XOR
B = B ^ A
print ("B = " , B)

##Rotacion de BITS
B = rotacionIz(B , 3)
print ("B = " , B)
```

```python
#Message DIGEST MD5 sha-256 sha-512
import hashlib
mensaje = "clase de Criptografia"
md5 = hashlib.md5(mensaje.encode()).hexdigest()
sha256 = hashlib.sha256(mensaje.encode()).hexdigest()
sha512 = hashlib.sha512(mensaje.encode()).hexdigest()

print("mensaje", mensaje)
print("MD5", md5)
print("sha-256", sha256)
print("sha-512", sha512)
```

```python
mensaje1 = "Criptografia momento1"
mensaje2 = "Criptografia momento2"
hash1 = hashlib.sha256(mensaje1.encode()).hexdigest
hash2 = hashlib.sha256(mensaje2.encode()).hexdigest
print("mensaje uno ", hash1)
print("mensaje dos ", hash2)
```

```python
## crear el hash_mensaje ....

from Crypto.PublicKey import RSA
from Crypto.Signature import pkcs1_15
from Crypto.Hash import SHA256

clavePri = RSA.generate(2048)
clavePu = clavePri.public_key()

mensaje = b"Documento autorizado"

hash_mensaje = SHA256.new(mensaje)

firma = pkcs1_15.new(clavePri).sign(hash_mensaje)

print("Mensaje" , mensaje.decode())
print("Firma digital" , firma.hex())
```

```python
try:
  pkcs1_15.new(clavePu).verify(
      SHA256.new(mensaje),
      firma
  )
  print("Firma Valida")

except (ValueError , TypeError):
  print("Firma no valida")
```

```python
mensaje_modificado = b"Documento no autorizado"
try:
  pkcs1_15.new(clavePu).verify(
      SHA256.new(mensaje_modificado),
      firma
  )
  print("Firma Valida")

except (ValueError , TypeError):
  print("Firma no valida")
```
